# Student Average Calculator - Documentation
## 1. Identify the components 

### 1.1 Inputs 
- 'quiz_1' (float) - mark for Quiz 1
- 'quiz_2' (float) - mark for Quiz 2
- 'quiz_3' (float) - mark for Quiz 3
- 'choice' (string) - "y" or "n" to continue 

### 1.2 Process
1. Initialize 'choice' to "y".
2. While 'choice' is "y":
   - Ask the user for 3 quiz marks.
   - Calculate the average (sum of marks divided by 3).
   - Print the average
   - Check if average is 50 or above.
   - Print  "Passed" if true , else "Failed".
   - Ask the user if they  want to continue.
3. Print "Program Ended" when the loop finishes.


## 1.3 Output
- The calculated average mark (displayed as a number).
- "Passed" or "Failed" message.
- "Program Ended" when the program finishes.

## 2. Design the Algorithm 
### 2.1 Pseudocode
START 
choice = "y"
WHILE choice == 'y"DO
INPUT quiz_1, quiz_2, quiz_3
average = (quiz_1 + quiz_2 + quiz_3)/3
PRINT average 
IF average >= 509 THEN
PRINT "Passed"
ELSE
PRITNT "Failed"
END IF 
PRINT "Continue? (y/n)" INPUT choice 
END WHILE 
PRINT "Program Ended"
END
